```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
```

```python
ord_pipeline = Pipeline(
    [
        ("Encoder" , OrdinalEncoder(categories= [["Low" , "Medium" , "High"]])),
        ("Scaler", StandardScaler())
    ]
)
```

```python
numeric_pipeline = Pipeline(
    [
        ("Scaler" , StandardScaler())
    ]
)
```

```python
preprocessor = ColumnTransformer(
    [
    ("categorical" , ord_pipeline , ["Range_Anxiety_Level"]),
    ("num" , numeric_pipeline , ["Age" , "Annual_Income_USD" , "Daily_Commute_km" , "Number_of_Cars_Owned" , "Charging_Stations_Near_Home" , "Charging_Stations_Near_Work" , "Environmental_Concern_Level" , "Home_Charging_Possible" , "Subsidy_Available"])
    ]
)
```

```python
vechile_df
```

```python
vechile_df["Will_Buy_EV"] = vechile_df["Will_Buy_EV"].map({"No" : 0 , "Yes" : 1})
vechile_df["Subsidy_Available"] = vechile_df["Subsidy_Available"].map({"Yes" : 1 , "No" : 0})
vechile_df["Home_Charging_Possible"] = vechile_df["Home_Charging_Possible"].map({"Yes" : 1 , "No" : 0})
```

```python
X = vechile_df.drop(columns = ["id" , "Gender" , "City_Type" , "Will_Buy_EV" , "Current_Car_Type"])
y = vechile_df["Will_Buy_EV"]
```

```python
X_sample , _ , y_sample , _ = train_test_split(
    X , y ,
    train_size=0.15,
    stratify=y,
    random_state=42
)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logistic_pipeline = Pipeline(
    [
        ("preprocessing" , preprocessor),
        ("model" , LogisticRegression(C=1.93 , max_iter=1000) )
    ]
)
```

```python
scores = cross_val_score(
    logistic_pipeline, X_sample, y_sample,
    cv=cv,
    scoring='roc_auc',
    n_jobs=2
)

print(f"AUC per fold: {scores}")
print(f"Mean AUC: {scores.mean():.4f} (+/- {scores.std():.4f})")
```

```python
from sklearn.svm import LinearSVC
```

```python
Svc_pipeline = Pipeline(
    [
        ("preprocessing" , preprocessor),
        ("model" , LinearSVC(C=100 , random_state=42) )
    ]
)
```

```python
scores = cross_val_score(
    Svc_pipeline, X_sample, y_sample,
    cv=cv,
    scoring='roc_auc',
    n_jobs=2
)

print(f"AUC per fold: {scores}")
print(f"Mean AUC: {scores.mean():.4f} (+/- {scores.std():.4f})")
```

```python
from sklearn.calibration import CalibratedClassifierCV
from sklearn.preprocessing import PolynomialFeatures
```

```python
Svc_poly_pipeline = Pipeline(
    [
        ("preprocessing", preprocessor),
        ("PolynomialFeatures", PolynomialFeatures(degree=2)),
        ("model", CalibratedClassifierCV(
            LinearSVC(tol=0.0001, max_iter=5000, loss='squared_hinge', C=10, random_state=42),
            cv=3
        ))
    ]
)
```

```python
scores = cross_val_score(
    Svc_poly_pipeline, X_sample, y_sample,
    cv=cv,
    scoring='roc_auc',
    n_jobs=2
)

print(f"AUC per fold: {scores}")
print(f"Mean AUC: {scores.mean():.4f} (+/- {scores.std():.4f})")
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
Tree_pipeline = Pipeline(
    [
        ("preprocessing" , preprocessor),
        ("model" , DecisionTreeClassifier(min_samples_split = 10, min_samples_leaf = 20, max_depth = 10, criterion = 'entropy' , random_state=42) )
    ]
)
```

```python
scores = cross_val_score(
    Tree_pipeline, X_sample, y_sample,
    cv=cv,
    scoring='roc_auc',
    n_jobs=2
)

print(f"AUC per fold: {scores}")
print(f"Mean AUC: {scores.mean():.4f} (+/- {scores.std():.4f})")
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
forest_pipeline = Pipeline(
    [
        ("preprocessing" , preprocessor),
        ("model" ,  RandomForestClassifier(n_estimators = 500, min_samples_split = 20, min_samples_leaf = 2, max_features = 'log2', max_depth = 20, criterion =  'entropy', random_state=42) )
    ]
)
```

```python
scores = cross_val_score(
    forest_pipeline, X_sample, y_sample,
    cv=cv,
    scoring='roc_auc',
    n_jobs=2
)

print(f"AUC per fold: {scores}")
print(f"Mean AUC: {scores.mean():.4f} (+/- {scores.std():.4f})")
```

```python
boosted_forest_pipeline = Pipeline(
    [
        ("Preprocessing" , preprocessor),
        ("Model" , XGBClassifier(subsample = 0.7, n_estimators = 200, min_child_weight = 5, max_depth = 4, learning_rate = 0.1, gamma = 1, colsample_bytree = 0.7 ,objective='binary:logistic', eval_metric='auc', random_state=42))
    ]
)
```

```python
scores = cross_val_score(
    boosted_forest_pipeline, X_sample, y_sample,
    cv=cv,
    scoring='roc_auc',
    n_jobs=2
)

print(f"AUC per fold: {scores}")
print(f"Mean AUC: {scores.mean():.4f} (+/- {scores.std():.4f})")
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
param_dist = {
    "Model__n_estimators": [100, 150, 200],
    "Model__max_depth": [2, 3, 4, 5],       
    "Model__learning_rate": [0.01, 0.05, 0.1],
    "Model__subsample": [0.6, 0.7, 0.8],
    "Model__colsample_bytree": [0.6, 0.7, 0.8],
    "Model__gamma": [0.5, 1, 2, 5],           
    "Model__min_child_weight": [5, 7, 10, 15] 
}
```

```python
random_boosted_search = RandomizedSearchCV(
    boosted_forest_pipeline,
    param_distributions=param_dist,
    n_iter=75,
    cv=5,
    scoring='roc_auc',
    random_state=42,
    n_jobs=2
)

random_boosted_search.fit( X_sample, y_sample)
print(random_boosted_search.best_params_)
print(random_boosted_search.best_score_)
```
