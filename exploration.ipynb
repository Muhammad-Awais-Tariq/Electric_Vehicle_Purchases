```python
ord_pipeline = Pipeline(
    [
        ("Encoder" , OrdinalEncoder(categories= [["Low" , "Medium" , "High"]])),
        ("Scaler", StandardScaler())
    ]
)
```

```python
numeric_pipeline = Pipeline(
    [
        ("Scaler" , StandardScaler())
    ]
)
```

```python
preprocessor = ColumnTransformer(
    [
    ("categorical" , ord_pipeline , ["Range_Anxiety_Level"]),
    ("num" , numeric_pipeline , ["Age" , "Annual_Income_USD" , "Daily_Commute_km" , "Number_of_Cars_Owned" , "Charging_Stations_Near_Home" , "Charging_Stations_Near_Work" , "Environmental_Concern_Level" , "Home_Charging_Possible" , "Subsidy_Available"])
    ]
)
```

```python
vechile_df
```

```python
vechile_df["Will_Buy_EV"] = vechile_df["Will_Buy_EV"].map({"No" : 0 , "Yes" : 1})
vechile_df["Subsidy_Available"] = vechile_df["Subsidy_Available"].map({"Yes" : 1 , "No" : 0})
vechile_df["Home_Charging_Possible"] = vechile_df["Home_Charging_Possible"].map({"Yes" : 1 , "No" : 0})
```

```python
X = vechile_df.drop(columns = ["id" , "Gender" , "City_Type" , "Will_Buy_EV" , "Current_Car_Type"])
y = vechile_df["Will_Buy_EV"]
```

```python
X_sample , _ , y_sample , _ = train_test_split(
    X , y ,
    train_size=0.15,
    stratify=y,
    random_state=42
)
```

```python
logistic_pipeline = Pipeline(
    [
        ("preprocessing" , preprocessor),
        ("model" , LogisticRegression(C=0.01 , max_iter=1000 ,solver = 'saga', penalty = 'elasticnet' , random_state=42) )
    ]
)
```

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scores = cross_val_score(
    logistic_pipeline, X, y,
    cv=cv,
    scoring='roc_auc',
    n_jobs=2
)

print(f"AUC per fold: {scores}")
print(f"Mean AUC: {scores.mean():.4f} (+/- {scores.std():.4f})")
```

```python
Svc_pipeline = Pipeline(
    [
        ("preprocessing" , preprocessor),
        ("model" , CalibratedClassifierCV(LinearSVC(C=100 , random_state=42 )) )
    ]
)
```

```python
scores = cross_val_score(
    Svc_pipeline, X_sample, y_sample,
    cv=cv,
    scoring='roc_auc',
    n_jobs=2
)

print(f"AUC per fold: {scores}")
print(f"Mean AUC: {scores.mean():.4f} (+/- {scores.std():.4f})")
```

```python
Svc_poly_pipeline = Pipeline(
    [
        ("preprocessing", preprocessor),
        ("PolynomialFeatures", PolynomialFeatures(degree=2)),
        ("model", CalibratedClassifierCV(
            LinearSVC(tol=0.0001, max_iter=5000, loss='squared_hinge', C=10, random_state=42),
            cv=3
        ))
    ]
)
```

```python
scores = cross_val_score(
    Svc_poly_pipeline, X_sample, y_sample,
    cv=cv,
    scoring='roc_auc',
    n_jobs=2
)

print(f"AUC per fold: {scores}")
print(f"Mean AUC: {scores.mean():.4f} (+/- {scores.std():.4f})")
```

```python
Tree_pipeline = Pipeline(
    [
        ("preprocessing" , preprocessor),
        ("model" , DecisionTreeClassifier(min_samples_split = 10, min_samples_leaf = 20, max_depth = 10, criterion = 'entropy' , random_state=42) )
    ]
)
```

```python
scores = cross_val_score(
    Tree_pipeline, X_sample, y_sample,
    cv=cv,
    scoring='roc_auc',
    n_jobs=2
)

print(f"AUC per fold: {scores}")
print(f"Mean AUC: {scores.mean():.4f} (+/- {scores.std():.4f})")
```

```python
forest_pipeline = Pipeline(
    [
        ("preprocessing" , preprocessor),
        ("model" ,  RandomForestClassifier(n_estimators = 500, min_samples_split = 20, min_samples_leaf = 2, max_features = 'log2', max_depth = 20, criterion =  'entropy', random_state=42) )
    ]
)
```

```python
stack_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

stacking_clf = StackingClassifier(
    estimators=[
        ("logistic", logistic_pipeline),
        ("cat", catboost_pipeline),
        ("rf", forest_pipeline),
    ],
    final_estimator=LogisticRegression(max_iter=1000, C=0.1, random_state=42),
    cv=stack_cv,
    n_jobs=1,
    passthrough=False,
)
```

```python
meta_param_grid = {
    "final_estimator__C": [0.01, 0.1, 1, 10],
    "final_estimator__class_weight": [None, "balanced"],
}

meta_cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

meta_search = GridSearchCV(
    estimator=stacking_clf,
    param_grid=meta_param_grid,
    scoring="roc_auc",
    cv=meta_cv,
    n_jobs=1,
    verbose=2,
    refit=True,
)

meta_search.fit(X, y)

print(meta_search.best_score_)
print(meta_search.best_params_)
```
